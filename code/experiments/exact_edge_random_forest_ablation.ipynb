```python
from pathlib import Path
import csv
import gc
import hashlib
import json
import random
import shutil
import subprocess
import sys
import time
from datetime import datetime, timezone

import numpy as np
import torch


def sha256_file(path):
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for chunk in iter(lambda: handle.read(8 * 1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def sha256_array(values, dtype):
    array = np.ascontiguousarray(values, dtype=dtype)
    return hashlib.sha256(array.tobytes(order="C")).hexdigest()


def write_csv(path, rows):
    assert rows, path
    with path.open("w", newline="", encoding="utf-8") as handle:
        writer = csv.DictWriter(handle, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)


assert hashlib.sha256(EMBEDDED_PROTOCOL_JSON.encode("utf-8")).hexdigest() == PROTOCOL_SHA256
embedded_protocol = json.loads(EMBEDDED_PROTOCOL_JSON)
assert embedded_protocol["status"] == "AMENDED AND REFROZEN AFTER V4 RESULTS AND BEFORE ANY V5 20-PARTITION RESULT"


try:
    import torch_geometric
except ModuleNotFoundError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "torch_geometric"])
    import torch_geometric

from scipy.stats import binomtest, pearsonr, spearmanr, t as student_t, wilcoxon
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    average_precision_score, confusion_matrix, f1_score,
    matthews_corrcoef, precision_score, recall_score, roc_auc_score,
)
from sklearn.model_selection import train_test_split
from torch_geometric.data import HeteroData


data_candidates = [
    path for path in Path("/kaggle/input").rglob("data.pt")
    if f"/{DATASET_NAME.lower()}/processed/data.pt" in str(path).replace("\\", "/").lower()
]
assert len(data_candidates) == 1, [str(path) for path in data_candidates]
data_path = data_candidates[0]
data_sha256 = sha256_file(data_path)
assert data_sha256 == SOURCE_DATA_SHA256, {
    "expected": SOURCE_DATA_SHA256, "observed": data_sha256,
}
loaded = torch.load(data_path, map_location="cpu", weights_only=False)
data = loaded[0] if isinstance(loaded, tuple) and len(loaded) == 2 else loaded
assert isinstance(data, HeteroData), type(data)
del loaded
gc.collect()

target_store = data[TARGET_NODE_TYPE]
assert target_store.x.ndim == 2 and target_store.x.shape[1] == 14
labeled_mask = (target_store.y == 0) | (target_store.y == 1)
labeled_ids = torch.where(labeled_mask)[0].cpu().numpy().astype(np.int64)
labeled_y = target_store.y[labeled_mask].cpu().numpy().astype(np.int64)
assert len(labeled_ids) == EXPECTED_LABELED
assert int((labeled_y == 1).sum()) == EXPECTED_POSITIVE
assert int((labeled_y == 0).sum()) == EXPECTED_NEGATIVE

ORIGINAL_NAMES = [
    "bal_c", "V_inv_c", "V_return_c", "N_inv_c", "N_return_c",
    "V_meaninv_c", "V_meanreturn_c", "bal_t", "V_inv_t", "V_return_t",
    "N_inv_t", "N_return_t", "V_meaninv_t", "V_meanreturn_t",
]
WEIGHTED_UFE_NAMES = [
    "call_in_rows", "call_out_rows", "call_in_count", "call_out_count",
    "call_in_sum", "call_out_sum", "trans_in_rows", "trans_out_rows",
    "trans_in_count", "trans_out_count", "trans_in_sum", "trans_out_sum",
]
STRUCTURAL_UFE_NAMES = ["call_in_rows", "call_out_rows", "trans_in_rows", "trans_out_rows"]
if EDGE_FEATURE_MODE == "weighted12":
    UFE_NAMES = WEIGHTED_UFE_NAMES
    TRANS_IN_SUM_COLUMN = 24
elif EDGE_FEATURE_MODE == "rows4":
    UFE_NAMES = STRUCTURAL_UFE_NAMES
    TRANS_IN_SUM_COLUMN = None
else:
    raise ValueError(EDGE_FEATURE_MODE)
FULL_FEATURE_NAMES = ORIGINAL_NAMES + UFE_NAMES
EXPECTED_FULL_DIM = 14 + len(UFE_NAMES)
RELATION_TO_INDEX = {"call": 0, "trans": 1}


def set_all_seeds(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def exact_full_edge_ufe(node_ids, chunk_size=2_000_000):
    """Exact target statistics from full edge_index used to construct model inputs."""
    node_ids = np.asarray(node_ids, dtype=np.int64)
    lookup = torch.full((target_store.num_nodes,), -1, dtype=torch.long)
    lookup[torch.from_numpy(node_ids)] = torch.arange(len(node_ids), dtype=torch.long)
    result = torch.zeros((len(node_ids), len(UFE_NAMES)), dtype=torch.float64)
    for edge_type in data.edge_types:
        source_type, relation, destination_type = edge_type
        assert relation in RELATION_TO_INDEX, relation
        store = data[edge_type]
        edge_count = int(store.edge_index.shape[1])
        if EDGE_FEATURE_MODE == "weighted12":
            assert hasattr(store, "edge_attr") and store.edge_attr.shape[1] >= 2
        for start in range(0, edge_count, chunk_size):
            stop = min(start + chunk_size, edge_count)
            if destination_type == TARGET_NODE_TYPE:
                positions = lookup[store.edge_index[1, start:stop]]
                keep = positions >= 0
                if bool(keep.any()):
                    selected = positions[keep]
                    if EDGE_FEATURE_MODE == "rows4":
                        column = RELATION_TO_INDEX[relation] * 2
                        result[:, column + 0].index_add_(0, selected, torch.ones(len(selected), dtype=torch.float64))
                    else:
                        column = RELATION_TO_INDEX[relation] * 6
                        attributes = torch.nan_to_num(
                            store.edge_attr[start:stop, :2][keep].to(torch.float64),
                            nan=0.0, posinf=0.0, neginf=0.0,
                        ).clamp_min_(0.0)
                        result[:, column + 0].index_add_(0, selected, torch.ones(len(selected), dtype=torch.float64))
                        result[:, column + 2].index_add_(0, selected, attributes[:, 0])
                        result[:, column + 4].index_add_(0, selected, attributes[:, 1])
            if source_type == TARGET_NODE_TYPE:
                positions = lookup[store.edge_index[0, start:stop]]
                keep = positions >= 0
                if bool(keep.any()):
                    selected = positions[keep]
                    if EDGE_FEATURE_MODE == "rows4":
                        column = RELATION_TO_INDEX[relation] * 2
                        result[:, column + 1].index_add_(0, selected, torch.ones(len(selected), dtype=torch.float64))
                    else:
                        column = RELATION_TO_INDEX[relation] * 6
                        attributes = torch.nan_to_num(
                            store.edge_attr[start:stop, :2][keep].to(torch.float64),
                            nan=0.0, posinf=0.0, neginf=0.0,
                        ).clamp_min_(0.0)
                        result[:, column + 1].index_add_(0, selected, torch.ones(len(selected), dtype=torch.float64))
                        result[:, column + 3].index_add_(0, selected, attributes[:, 0])
                        result[:, column + 5].index_add_(0, selected, attributes[:, 1])
    del lookup
    return torch.log1p(result).numpy()


def materialize_exact_ids(node_ids):
    """Materialize deterministic base+UFE features from every incident full-graph edge."""
    started = time.perf_counter()
    ids = np.sort(np.asarray(node_ids, dtype=np.int64))
    assert len(ids) == len(np.unique(ids))
    ids_tensor = torch.from_numpy(ids)
    base = target_store.x[ids_tensor].cpu().numpy().astype(np.float64, copy=True)
    labels = target_store.y[ids_tensor].cpu().numpy().astype(np.int64, copy=True)
    assert set(np.unique(labels)) == {0, 1}
    exact_ufe_log = exact_full_edge_ufe(ids)
    features = np.concatenate([base, exact_ufe_log], axis=1).astype(np.float64, copy=False)
    assert features.shape == (len(ids), EXPECTED_FULL_DIM)
    assert np.isfinite(features).all()
    return ids, features, labels, time.perf_counter() - started


def rows_for_ids(all_ids, all_features, all_labels, requested_ids):
    requested_ids = np.sort(np.asarray(requested_ids, dtype=np.int64))
    positions = np.searchsorted(all_ids, requested_ids)
    assert np.array_equal(all_ids[positions], requested_ids)
    return (
        all_ids[positions].copy(),
        all_features[positions].copy(),
        all_labels[positions].copy(),
    )


def feature_view(full_features, variant, shuffle_seed=None):
    assert full_features.shape[1] == EXPECTED_FULL_DIM
    if variant == "base14":
        return full_features[:, :14].copy()
    if variant == "ufe12_only":
        return full_features[:, 14:].copy()
    if variant == "base14_ufe12":
        return full_features.copy()
    if variant == "trans_in_sum_only":
        return full_features[:, [TRANS_IN_SUM_COLUMN]].copy()
    if variant == "base14_trans_in_sum":
        return np.concatenate([full_features[:, :14], full_features[:, [TRANS_IN_SUM_COLUMN]]], axis=1)
    if variant == "base14_ufe11_without_trans_in_sum":
        keep = list(range(14)) + [column for column in range(14, 26) if column != TRANS_IN_SUM_COLUMN]
        return full_features[:, keep].copy()
    if variant == "base14_shuffled_ufe12":
        assert shuffle_seed is not None
        permutation = np.random.default_rng(shuffle_seed).permutation(len(full_features))
        return np.concatenate([full_features[:, :14], full_features[permutation, 14:]], axis=1)
    if variant == "rows4_only":
        return full_features[:, 14:].copy()
    if variant == "base14_rows4":
        return full_features.copy()
    if variant == "base14_in_rows2":
        return np.concatenate([full_features[:, :14], full_features[:, [14, 16]]], axis=1)
    if variant == "base14_out_rows2":
        return np.concatenate([full_features[:, :14], full_features[:, [15, 17]]], axis=1)
    if variant == "base14_shuffled_rows4":
        assert shuffle_seed is not None
        permutation = np.random.default_rng(shuffle_seed).permutation(len(full_features))
        return np.concatenate([full_features[:, :14], full_features[permutation, 14:]], axis=1)
    raise ValueError(variant)


SEARCH_SPACE = [
    {"n_estimators": n, "max_depth": depth}
    for n in (100, 200, 300) for depth in (None, 10)
]


def build_model(config, seed):
    return RandomForestClassifier(
        n_estimators=config["n_estimators"],
        max_depth=config["max_depth"],
        min_samples_leaf=1,
        class_weight=None,
        random_state=seed,
        n_jobs=-1,
    )


def positive_scores(model, features):
    probabilities = model.predict_proba(features)
    positive_column = int(np.where(model.classes_ == 1)[0][0])
    return probabilities[:, positive_column]


def metric_bundle(labels, scores):
    labels = np.asarray(labels, dtype=np.int64)
    scores = np.asarray(scores, dtype=np.float64)
    predictions = (scores >= THRESHOLD).astype(np.int64)
    tn, fp, fn, tp = confusion_matrix(labels, predictions, labels=[0, 1]).ravel()
    return {
        "pr_auc": float(average_precision_score(labels, scores)),
        "roc_auc": float(roc_auc_score(labels, scores)),
        "precision": float(precision_score(labels, predictions, zero_division=0)),
        "recall": float(recall_score(labels, predictions, zero_division=0)),
        "binary_f1": float(f1_score(labels, predictions, zero_division=0)),
        "micro_f1": float(f1_score(labels, predictions, average="micro")),
        "macro_f1": float(f1_score(labels, predictions, average="macro")),
        "mcc": float(matthews_corrcoef(labels, predictions)),
        "fpr": float(fp / (fp + tn)),
        "tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp),
    }


def select_winner(x_train, y_train, x_validation, y_validation, tuning_seeds):
    raw = []
    for config_id, config in enumerate(SEARCH_SPACE):
        for seed in tuning_seeds:
            set_all_seeds(seed)
            model = build_model(config, seed)
            started = time.perf_counter()
            model.fit(x_train, y_train)
            fit_seconds = time.perf_counter() - started
            metrics = metric_bundle(y_validation, positive_scores(model, x_validation))
            raw.append({
                "config_id": config_id,
                "config_json": json.dumps(config, sort_keys=True),
                "seed": seed,
                "fit_seconds": fit_seconds,
                **{f"validation_{key}": value for key, value in metrics.items()},
            })
            del model
    summaries = []
    for config_id, config in enumerate(SEARCH_SPACE):
        rows = [row for row in raw if row["config_id"] == config_id]
        summaries.append({
            "config_id": config_id,
            "config": config,
            "mean_macro_f1": float(np.mean([row["validation_macro_f1"] for row in rows])),
            "mean_fpr": float(np.mean([row["validation_fpr"] for row in rows])),
            "mean_pr_auc": float(np.mean([row["validation_pr_auc"] for row in rows])),
            "mean_fit_seconds": float(np.mean([row["fit_seconds"] for row in rows])),
        })
    winner = max(summaries, key=lambda row: (
        row["mean_macro_f1"], -row["mean_fpr"], row["mean_pr_auc"], -row["config_id"],
    ))
    return winner, raw, summaries


def holm_adjust(p_values):
    p_values = np.asarray(p_values, dtype=float)
    order = np.argsort(p_values)
    adjusted = np.empty_like(p_values)
    running = 0.0
    m = len(p_values)
    for rank, index in enumerate(order):
        running = max(running, (m - rank) * p_values[index])
        adjusted[index] = min(1.0, running)
    return adjusted


timestamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
run_dir = Path("/kaggle/working") / f"{RUN_SLUG}_{timestamp}"
run_dir.mkdir(parents=True, exist_ok=False)
(run_dir / "ablation_ufe_verification_protocol_v5_20splits.json").write_text(
    EMBEDDED_PROTOCOL_JSON, encoding="utf-8"
)
assert sha256_file(run_dir / "ablation_ufe_verification_protocol_v5_20splits.json") == PROTOCOL_SHA256
print("Protocol SHA256:", PROTOCOL_SHA256)
print("Dataset:", DATASET_NAME, data_path)
print("data.pt SHA256:", data_sha256)
print("Target:", TARGET_NODE_TYPE, "labeled:", EXPECTED_LABELED)
print("Output:", run_dir)
print("Torch/PyG:", torch.__version__, torch_geometric.__version__)
```
